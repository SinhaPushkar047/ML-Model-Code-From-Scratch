```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import r2_score
```

# **Using Linear Regression Library**

```python
from sklearn.datasets import load_diabetes
data=load_diabetes()
```

```python
from sklearn.model_selection import train_test_split
x = data.data
y = data.target
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42
)
```

```python
x.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```

# **Using Gradient Descent**

```python
class GD:
  def __init__(self,lr,epoch):
    self.m=100
    self.b=np.ones(x.shape[1]+1)
    self.lr=lr
    self.epoch=epoch
  def fit(self,x,y):
    x=np.insert(x,0,1,axis=1)
    for i in range(self.epoch):
      loss_b=(-2/x.shape[0])*(x.T.dot((y.ravel()-x.dot(self.b))))
      self.b=self.b-self.lr*loss_b
    return(self.b[0],self.b[1:])
  def pred(self,x):
    x=np.insert(x,0,1,axis=1)
    return x.dot(self.b)
```

```python
mygd=GD(0.5,1000000)
```

```python
mygd.fit(x_train,y_train)
```

```python
y_pred_lr=mygd.pred(x_test)
```

```python
r2_score(y_test,y_pred_lr)
```
